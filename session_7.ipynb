```python
import pandas as pd 

df=pd.read_csv('zomato.csv')
col='rate'
df['rate'] = df['rate'].str.replace('/5', '', regex=False)
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')

q1 = df[col].quantile(0.25)
q3 = df[col].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
outliers.index
```

```python
import matplotlib.pyplot as plt 
df=pd.read_csv('ecommerce_delivery_analytics.csv')
plt.boxplot(df['Order Value (INR)'])
plt.xlabel('Orders')
plt.ylabel('Order Amount (INR)')
plt.title('Boxplot of Order Amount')
plt.show()
```

```python
df=pd.read_csv('digital_wallet_transactions.csv')
df
lower = df['transaction_fee'].quantile(0.05)
upper = df['transaction_fee'].quantile(0.95)

df['transaction_fee'] = df['transaction_fee'].clip(lower, upper)
print(df['transaction_fee'].describe())
```
