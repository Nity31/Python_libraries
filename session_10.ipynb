```python
import matplotlib.pyplot as plt
import numpy as np


days=np.array([1,2,3,4,5,6,7])
steps=np.array([2000,5000,6000,7000,75000,8000,9000])
plt.plot(days,steps)

plt.savefig('steps_lineplot.png')
plt.show()
```

```python
ratings=np.array([4,3,2,5,1,4,5,3,4,1])
price=np.array([220,320,340,450,550,660,670,700,750,760])

plt.scatter(ratings,price)
plt.title('Restaurant Ratings and price')
plt.xlabel('Zomato Ratings',fontdict={'size':12,'color':'black'})
plt.ylabel('Meal price',fontdict={'size':12,'color':'black'})
plt.show()
```

```python
apps = np.array(["Swiggy", "Zomato", "Domino's"])
orders = np.array([15, 10, 7])

plt.bar(apps , orders, color=['red','orange','blue'],label='orders')
plt.title("orders placed last months")
plt.xlabel('food delivery apps')
plt.ylabel('number of orders')
plt.legend()
plt.show()

```

```python
durations = np.array([18, 25, 32, 40, 22,35, 28, 45, 30, 38,50, 27, 33, 41, 29,36, 24, 47, 31, 39])

plt.hist(durations,bins=5,color='red')
plt.show()
```

```python
weeks=np.array(['m','t','w','thr','f','sat','sun'])
screen_time=np.array([5,7,8,10,14,12,15])
post_likes=np.array([23,35,45,55,65,76,87])

fig, ax = plt.subplots(1, 2, figsize=(8,4))


ax[0].plot(weeks, screen_time)
ax[0].set_title("Weekly Instagram Screen Time")
ax[0].set_xlabel("Week Days")
ax[0].set_ylabel("Screen Time (Hours)")


ax[1].bar(weeks, post_likes)
ax[1].set_title("Weekly Instagram Likes")
ax[1].set_xlabel("Week Days")
ax[1].set_ylabel("Likes")

plt.tight_layout()
plt.savefig('social_media_usage.png')
plt.show()
```
