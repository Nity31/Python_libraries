```python
import numpy as np 

#1
friend_1=np.array([1200,2300,10000,5000,6000,2300,3400])
friend_2=np.array([5000,2000,23500,8000,7000,4500,1200])

print("Addition:",np.add(friend_1,friend_2))
print("Subtraction:",np.subtract(friend_1,friend_2))
print("Multiplication:",np.multiply(friend_1,friend_2))
print("Division:",np.divide(friend_1,friend_2))

#2
user_prefrences=np.array([[23,12,34],
                          [23,12,34],
                          [23,54,34]])

song_popularity=np.array([[3,2,5],
                          [4,5,1],
                          [2,3,5]])

print("User_prefrence:\n",user_prefrences)
print("song popularity:\n",song_popularity)

print("Dot Product:\n",np.dot(user_prefrences,song_popularity))
print("Matrix Multiplication:\n",np.matmul(user_prefrences,song_popularity))

#For 2D matrices, np.dot() and np.matmul() give the same result;
#the difference is only in how they handle higher-dimensional arrays.

#3
matrix=np.arange(1,17).reshape(4,4)
print(matrix)
transposed_matrix=matrix.T
print("Transposed Matrix:\n",transposed_matrix)
print("Mean:",np.mean(transposed_matrix))
print("Median:",np.median(transposed_matrix))
print("Standard Deviation :",np.std(transposed_matrix))
print("Variance :",np.var(transposed_matrix))

#4
matrix=np.array([[1,2,3],
                 [0,1,4],
                 [5,6,0]])
print("Original Matrix:\n",matrix)
print("Inverse of Matrix:\n",np.linalg.inv(matrix))
print("Determinant of Matrix:",np.linalg.det(matrix))
print("Eigenvalues and Eigenvectors of Matrix:\n",np.linalg.eig(matrix))

#5
arr=np.array([[120,500,234,222,122,223],
              [450,340,120,340,510,230]])
print(arr)
print("Reshaped array:\n",arr.reshape(3,4))
flattened_array=arr.flatten()
print("Flatten:",flattened_array)
parts=np.split(flattened_array,2)
print("Splitted array:",parts)
print("Vertical Stacking:",np.vstack(parts))
```
