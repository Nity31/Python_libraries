```python
import pandas as pd
import sweetviz as sv


mumbai = pd.read_csv("Zomato_Mumbai_Dataset.csv", on_bad_lines="skip")
delhi = pd.read_csv("DelhiNCR Restaurants.csv")

report = sv.compare([mumbai, "Mumbai"], [delhi, "Delhi"])

report.show_html("zomato_mumbai_delhi_report.html")
```
