```python
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset("tips")
sns.pairplot(df)
plt.show()
```

```python
df = sns.load_dataset("flights")

pivot_data = df.pivot_table(
    index="month",
    columns="year",
    values="passengers"
)

plt.figure(figsize=(12, 7))

sns.heatmap(
    pivot_data,
    annot=True,
    cmap="YlOrRd"
)

plt.title("Monthly Passenger Counts by Year")
plt.xlabel("Year")
plt.ylabel("Month")

plt.show()
```

```python
df = sns.load_dataset("fmri")

sns.relplot(
    data=df,
    x="timepoint",
    y="signal",
    hue="event",
    kind="line"
)

plt.title("FMRI Signal Changes Over Time")
plt.xlabel("Time Point")
plt.ylabel("Signal")

plt.show()
```

```python
df = sns.load_dataset("titanic")

sns.catplot(
    data=df,
    x="class",
    y="survived",
    kind="bar",
    errorbar="ci"
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

```python
df = sns.load_dataset("penguins")

sns.jointplot(
    data=df,
    x="bill_length_mm",
    y="flipper_length_mm",
    kind="reg"
)

plt.show()
```
