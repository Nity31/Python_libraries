```python
'''You have two CSV files: 'orders.csv' (order_id, user_id, amount) and 'users.csv' 
(user_id, username). 
    Load both into DataFrames using pathlib for file paths, 
    then merge them on 'user_id' to show a combined table with username and amount.'''
   
from pathlib import Path 
orders_data = {
    "order_id": [101, 102, 103, 104],
    "user_id": [1, 2, 1, 3],
    "amount": [500, 750, 300, 900]
}
orders = pd.DataFrame(orders_data)
orders.to_csv("orders.csv", index=False)

users_data = {
    "user_id": [1, 2, 3],
    "username": ["Rahul", "Priya", "Amit"]
}
users = pd.DataFrame(users_data)
users.to_csv("users.csv",index=False)

orders_path=Path('orders.csv')
users_path=Path('users.csv')

merge=pd.merge(orders,users,on='user_id')
merge
```

```python
today_orders = pd.DataFrame({
    "order_id": [101, 102],
    "item": ["Pizza", "Burger"],
    "price": [250, 150]
})
yesterday_orders = pd.DataFrame({
    "order_id": [99, 100],
    "item": ["Sandwich", "Pasta"],
    "price": [100, 200]
})

combined=pd.concat([today_orders,yesterday_orders]) .reset_index(drop=True)
combined

```
