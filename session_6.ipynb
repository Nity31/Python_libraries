```python
import pandas as pd 

df=pd.read_csv("https://raw.githubusercontent.com/12345k/IPL-Dataset/master/IPL/data.csv")
df
print("Data having Null values:",df.isnull().sum())
```

```python
print("Data having not Null Val",df.notnull().sum())
```

```python
print("Shape before dropping the missing values",df.shape)

df.dropna(axis=0 , how='any')
df
print("Shape after dropping the missing values",df.shape)

df.isnull().sum()
```

```python
df=pd.read_csv("zomato.csv")
df.drop("url",axis=1,inplace=True)
df
```

```python
df['rate']=df['rate'].bfill()
df

df.isnull().sum()

```

```python
df['rate']=df['rate'].ffill() 
df

df.isnull().sum()
```

```python
df2=pd.read_csv('zomato.csv')
df2.drop(columns='url')

print("Before filling null values :\n",df2.isnull().sum())
df2['rate']=df2['rate'].bfill()
print("After filling null values :\n",df2.isnull().sum())
```
