```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(10)
delivery_times = np.random.randint(20, 61, 50)

sns.histplot(delivery_times, bins=8, kde=True)

plt.title("Distribution of Zomato Delivery Times")
plt.xlabel("Delivery Time (Minutes)")
plt.ylabel("Number of Orders")

plt.show()
```

```python
df = sns.load_dataset("tips")

sns.boxplot(x="day", y="total_bill", data=df)
plt.title("Distribution of Total Bill by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill ($)")

plt.show()
```

```python
sns.set_theme(style="darkgrid")

teams = [
    "CSK", "MI", "RCB", "KKR",
    "RR", "SRH", "DC", "PBKS"
]

np.random.seed(10)

scores = {
    "Team": np.repeat(teams, 20),
    "Runs": np.random.randint(120, 220, 160)
}

df = pd.DataFrame(scores)

sns.violinplot(x="Team", y="Runs", data=df)

plt.title("IPL Team Run Distribution")
plt.xlabel("IPL Teams")
plt.ylabel("Runs Scored")

plt.show()
```

```python
genres = [
    "Pop", "Rock", "Hip-Hop", "Pop", "EDM",
    "Rock", "Pop", "Jazz", "Hip-Hop", "EDM",
    "Pop", "Rock", "Jazz", "Pop", "Hip-Hop",
    "EDM", "Rock", "Pop", "Jazz", "Hip-Hop",
    "Pop", "EDM", "Rock", "Pop", "Hip-Hop",
    "Jazz", "EDM", "Pop", "Rock", "Hip-Hop",
    "Pop", "Jazz", "EDM", "Rock", "Pop", "Hip-Hop",
    "Jazz", "EDM", "Pop", "Rock"
]

df = pd.DataFrame({"Genre": genres})

sns.countplot(x="Genre", data=df)

plt.title("Number of Spotify Tracks per Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")

plt.show()
```

```python
steps = [6500, 7200, 8000, 5500, 9000, 7500, 10000]

sns.set_theme(style="whitegrid")

sns.kdeplot(steps,fill=True,color="green")
plt.title("Distribution of Daily Step Counts")
plt.xlabel("Steps per Day")
plt.ylabel("Density")

plt.show()
```
