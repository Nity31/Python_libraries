```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 45, 50]

products = ["A", "B", "C", "D"]
quantities = [25, 40, 30, 20]

age = [18, 20, 22, 24, 26, 28]
marks = [55, 60, 65, 70, 75, 80]

categories = ["Food", "Clothes", "Electronics", "Books"]
expenses = [40, 25, 20, 15]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(months, sales, marker="o")
ax[0, 0].set_title("Monthly Sales")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales")


ax[0, 1].bar(products, quantities)
ax[0, 1].set_title("Product Quantities")
ax[0, 1].set_xlabel("Product")
ax[0, 1].set_ylabel("Quantity")

ax[1, 0].scatter(age, marks)
ax[1, 0].set_title("Age vs Marks")
ax[1, 0].set_xlabel("Age")
ax[1, 0].set_ylabel("Marks")

ax[1, 1].pie(expenses, labels=categories, autopct="%1.1f%%")
ax[1, 1].set_title("Expense Distribution")

plt.tight_layout()

plt.show()
```

```python
companies = ["Zomato", "Swiggy", "Domino's"]
delivery_time = [35, 32, 28]

plt.figure(figsize=(8, 5))

plt.bar(companies[0], delivery_time[0],
        color="red", edgecolor="black",
        linestyle="--", linewidth=2)

plt.bar(companies[1], delivery_time[1],
        color="orange", edgecolor="black",
        linestyle="-.", linewidth=3)

plt.bar(companies[2], delivery_time[2],
        color="blue", edgecolor="black",
        linestyle=":", linewidth=2)

plt.title("Average Delivery Time Comparison")
plt.xlabel("Food Delivery Platform")
plt.ylabel("Average Delivery Time (minutes)")

plt.show()
```

```python
influencers = ["Influencer A", "Influencer B", "Influencer C",
               "Influencer D", "Influencer E"]

followers = [2.5, 5.0, 3.5, 7.0, 4.0]     
daily_posts = [2, 4, 3, 5, 2.5]              


fig, ax1 = plt.subplots(figsize=(10, 6))


ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Followers (millions)", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Instagram Followers vs Average Daily Posts")

plt.tight_layout()
plt.show()
```

```python
movies = ["Dunki", "Animal", "Jawan", "12th Fail", "Stree 2"]

tickets_sold = [120, 180, 200, 80, 160]   # in lakhs
imdb_rating = [6.5, 6.1, 7.0, 8.8, 6.5]

plt.figure(figsize=(10, 6))

plt.scatter(
    tickets_sold,
    imdb_rating,
    color="blue",
    marker="o",
    s=100
)

plt.annotate("Dunki", (120, 6.5), xytext=(5, 8),
             textcoords="offset points")

plt.annotate("Animal", (180, 6.1), xytext=(5, 8),
             textcoords="offset points")

plt.annotate("Jawan", (200, 7.0), xytext=(5, 8),
             textcoords="offset points")

plt.annotate("12th Fail", (80, 8.8), xytext=(5, 8),
             textcoords="offset points")

plt.annotate("Stree 2", (160, 6.5), xytext=(5, 8),
             textcoords="offset points")

plt.xlabel("Tickets Sold (lakhs)")
plt.ylabel("IMDb Rating")
plt.title("Tickets Sold vs IMDb Ratings of Bollywood Movies")

plt.grid(True)
plt.show()
```
